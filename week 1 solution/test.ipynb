```python
# Write your load_fits function here.
import numpy as np
from astropy.io import fits
import matplotlib.pyplot as plt

def load_fits(image):
  hdulist = fits.open(image)
  df = hdulist[0].data
  return np.unravel_index(np.argmax(df), df.shape)


if __name__ == '__main__':
  # Run your `load_fits` function with examples:
  bright = load_fits('image1.fits')
  print(bright)

```

```python
hdulist = fits.open('image1.fits')
data = hdulist[0].data

  # Plot the 2D image data
plt.imshow(data.T, cmap = 'viridis')
plt.colorbar()
plt.show()
```

```python
# and use np.percentile to clip background noise and enhance faint details

hdulist = fits.open('image1.fits')
data = hdulist[0].data

  # Plot the 2D image data
plt.imshow(data.T,
           vmin=np.percentile(data, 10),
           vmax=np.percentile(data, 99.5)
          )
plt.colorbar()
plt.show()
```

```python
plt.figure(figsize=(8, 8))

# Use cmap='gray' so it's black-and-white, 
plt.imshow(
    data.T,
    cmap="gray",
    vmin=np.percentile(data, 10),
    vmax=np.percentile(data, 99.5),
    
)

plt.colorbar(label="Pixel Intensity")
plt.title("ESA Astronomical Image")
plt.show()
```
